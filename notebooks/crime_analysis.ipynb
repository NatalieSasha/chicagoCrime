# Crime Data Analysis and Visualization

The default CSV is a dated public snapshot of Chicago incident reports. The smaller sample CSV is synthetic and is provided only as a fallback.

```python
from pathlib import Path
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from src.data_loader import load_csv
from src.preprocessing import preprocess_data
from src.eda import overview, category_frequencies, temporal_counts
from src.insights import generate_insights
from src.visualizations import create_charts

DATA_PATH = os.getenv('CRIME_DATA_PATH', 'data/chicago_crime_2026_ytd.csv')
raw = load_csv(DATA_PATH)
raw.head()
```

## Understand and clean the data

```python
raw.info()
raw.isna().sum()
clean, preprocessing_report = preprocess_data(raw)
preprocessing_report
clean.describe(include='all')
```

## Exploratory and statistical analysis

```python
overview(clean)
category_frequencies(clean)
temporal_counts(clean, 'M')
if {'latitude', 'longitude'}.issubset(clean.columns):
    clean[['latitude', 'longitude']].describe()
numeric = clean.select_dtypes('number')
numeric.corr() if len(numeric.columns) > 1 else 'No pair of numeric fields for correlation.'
```

## Visualizations and findings

```python
create_charts(clean, 'outputs')
for observation in generate_insights(clean):
    print(observation)
# Counts describe only records in this dataset, not population rates.
# Coverage, reporting behavior, missingness, and population differences limit interpretation.
```

## Limitations and conclusion

This workflow is descriptive, not predictive or intended for law-enforcement decisions. Findings depend on the supplied dataset's coverage, definitions, and quality. This partial-year snapshot supports descriptive exploration but not full-year totals, seasonal conclusions, or population-adjusted rates.
